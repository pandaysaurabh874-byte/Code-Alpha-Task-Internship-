```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import html5lib
```

```python
url = "https://codewithharry.com"
```

```python
r= requests.get(url)
```

```python
htmlcontent = r.content
```

```python
print(r.content)
```

```python
soup = BeautifulSoup(r.content, 'html.parser')
```

```python
print(soup)
print(soup.prettify())
```

```python

title = soup.title
print(type(title))#tag
print(type(soup))#soup
print((type(title.string)))#string
```

```python
#Get the all the paragraphs from the page
paras = soup.find_all('p')
print(paras)
```

```python
#get all the anchors tags from the page 
anchors = soup.find_all('a')
print(anchors)
```

```python
#Get the first elememt in the HTML page...
print(soup.find('p'))


#Get classes of any element in the page 
print(soup.find('p').get('class'))


#Find all the elements with class lead
print(soup.find_all("p", class_="lead"))
```

```python
#Get the text from the tags/soup
print(soup.find('p').get_text())
print(soup.get_text())
```
